```python
import sys
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window
from pyspark.sql.types import StructType,StructField,StringType,IntegerType,DoubleType

spark = (SparkSession.builder
         .appName("check-python")
         .master("local[*]")
         .config("spark.pyspark.python", sys.executable)
         .config("spark.pyspark.driver.python", sys.executable)
         .getOrCreate())
```

**replace()**

Ещё одним способом работы со значениями в датафрейме является функция replace().

В PySpark функция replace используется для замены значений в DataFrame. Эта функция достаточно гибкая и может работать как со строками, так и с числовыми значениями. 

Вот пошаговое объяснение с примерами:

**Замена строки:**

Замена конкретной строки в столбце на другую строку.

```python
data = [("Алиса", "NY", 25),
        ("Боб", "CA", 30),
        ("Чарли", "CA", 35)]

columns = ["Имя", "Штат", "Возраст"]

df = spark.createDataFrame(data, columns)
df.show(truncate=False)
```

```python

# Замена "CA" на "Калифорния" в столбце "Штат"
df = df.replace("CA", "Калифорния", "Штат")
df.show(truncate=False)
```

**Замена нескольких значений:**

```python
# Замена "CA" на "Калифорния" и "NY" на "Нью-Йорк"
df = df.replace(["CA", "NY"], ["Калифорния", "Нью-Йорк"], "Штат")

df.show(truncate=False)
```

**Замена с использованием словаря:**

Для сложных замен можно использовать словарь, который сопоставляет старые значения новым.

```python
# Замена по словарю
replacement_dict = {"CA": "Калифорния", "NY": "Нью-Йорк"}
df = df.replace(replacement_dict, subset=["Штат"])

df.show()
```
