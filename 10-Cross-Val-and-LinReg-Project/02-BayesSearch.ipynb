```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../DATA/Advertising.csv")
```

```python
## Создаём X и y
X = df.drop('sales',axis=1)
y = df['sales']

# Разбиение на обучающий и тестовый наборы - TRAIN TEST SPLIT
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)

# Масштабирование данных (SCALE)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import ElasticNet

base_elastic_model = ElasticNet()
```

```python
param_grid = {'alpha':[0.1,1,5,10,50,100],
              'l1_ratio':[.1, .5, .7, .9, .95, .99, 1]}
```

```python
from skopt import BayesSearchCV

bayes_model = BayesSearchCV(estimator=base_elastic_model,
                          search_spaces=param_grid,
                          scoring='neg_mean_squared_error',
                          cv=5,
                          verbose=0)

import time

start = time.ctime()

bayes_model.fit(X_train,y_train)

end = time.ctime()

print(f'Time spent: {int(end[-7:-5])-int(start[-7:-5])} sec')
```

```python
bayes_model.best_params_
```

```python
from skopt import BayesSearchCV
from skopt.space import Real, Integer, Categorical

search_spaces = {'alpha': Real(0.1,100),
              'l1_ratio':Real(.1, 1)}

bayes_model = BayesSearchCV(estimator=base_elastic_model,
                          search_spaces=search_spaces,
                          scoring='neg_mean_squared_error',
                          cv=5,
                          verbose=0,
                          n_jobs=-1)

import time

start = time.ctime()

bayes_model.fit(X_train,y_train)

end = time.ctime()

print(f'Time spent: {int(end[-7:-5])-int(start[-7:-5])} sec')
```

```python
bayes_model.best_params_
```
