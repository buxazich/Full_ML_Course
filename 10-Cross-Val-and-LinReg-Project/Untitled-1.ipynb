```python
import pandas as pd 
import statsmodels.api as sm 

data = {'x': (2,2,1,4,5),
        'y': (2,2,3,4,5)}

mdat = pd.DataFrame(data=data)
print('\n', mdat)
```

```python
o1 = sm.OLS.from_formula('y~x', data=mdat).fit()
print('\n', o1.summary())
```

```python
import seaborn as sns 

sns.lmplot(y='y',x='x',data=mdat, aspect=1.2, height=2)
```
